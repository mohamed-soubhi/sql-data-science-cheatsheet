## Setup — build the sample database this notebook runs on
*Run the cell below first: it creates an in-memory SQLite database and the `q()` helper that prints every query's result set.*

````python
import sqlite3
import textwrap

conn = sqlite3.connect(':memory:')  # throw-away in-memory database
conn.execute("PRAGMA foreign_keys = ON")

def q(sql):
    """Run a query, print formatted SQL, and display the result set as a table."""
    clean_sql = textwrap.dedent(sql).strip()

    # 1. Print formatted SQL query
    print("\n" + "=" * 60)
    print("QUERY:")
    print(textwrap.indent(clean_sql, "  "))
    print("-" * 60)

    cur = conn.execute(sql)
    rows = cur.fetchall()

    if cur.description:  # SELECT / statement returning rows
        headers = [d[0] for d in cur.description]
        str_rows = [[("" if v is None else str(v)) for v in r] for r in rows]

        # Calculate dynamic column widths for clean alignment
        widths = [
            max(len(h), *(len(r[i]) for r in str_rows), 1)
            for i, h in enumerate(headers)
        ] if str_rows else [len(h) for h in headers]

        header_line = " | ".join(h.ljust(widths[i]) for i, h in enumerate(headers))
        sep_line = "-+-".join("-" * widths[i] for i in range(len(headers)))

        print(header_line)
        print(sep_line)
        for r in str_rows:
            print(" | ".join(val.ljust(widths[i]) for i, val in enumerate(r)))
        if not str_rows:
            print("(0 rows returned)")
    else:  # DDL / DML (CREATE, INSERT, UPDATE, DELETE)
        affected = cur.rowcount
        print(f"Done (rows affected: {affected})" if affected != -1 else "Done.")

    print("=" * 60)

print('Connection ready:', conn)
````

## 📗 Key definitions

- **Data** — a collection of facts (words, numbers, pictures). A business's most critical asset.
- **Database** — a repository of data; supports adding, modifying, querying.
- **DBMS/RDBMS** — the software tools controlling access, organization, storage (MySQL, Oracle, DB2, SQLite…).
- **SQL** — Structured Query Language: how we ask the database for what we want.

## 🧪 The five basic commands — all of them, live

**1️⃣ CREATE** — build a table with named columns and datatypes:
```sql
CREATE TABLE people (id INTEGER PRIMARY KEY, name TEXT, age INTEGER, city TEXT)
```
*Run the next cell, then each following one.*

````python
q("CREATE TABLE people (id INTEGER PRIMARY KEY, name TEXT, age INTEGER, city TEXT)")
````

**2️⃣ INSERT** — populate the table (Create):

````python
q("INSERT INTO people VALUES (1, 'Raul', 41, 'Toronto')")
q("INSERT INTO people VALUES (2, 'Rav', 35, 'Markham')")
q("INSERT INTO people VALUES (3, 'Emma', 29, 'Sydney')")
q("SELECT * FROM people")
````

**3️⃣ SELECT** — read (Read). Restrict with **WHERE** and choose **columns**:

````python
q("SELECT name, city FROM people WHERE age > 30")
````

**4️⃣ UPDATE** — change a row (Update). The WHERE clause picks *which* row:

````python
q("UPDATE people SET city = 'Ottawa' WHERE id = 1")
q("SELECT * FROM people")
````

**5️⃣ DELETE** — remove a row (Delete):

````python
q("DELETE FROM people WHERE id = 3")
q("SELECT * FROM people")
````

## 💡 Tips & tricks

- These five commands are **identical on every RDBMS** (MySQL, Oracle, DB2, Postgres…).
  Dialects differ on advanced stuff, not on CRUD.
- `database ≈ DBMS ≈ database server` — the terms are used interchangeably in practice.
